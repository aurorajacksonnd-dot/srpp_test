```python
import pandas
import numpy
```

```python
df = pandas.read_csv("GTEx_Analysis_2025-08-22_v11_RNASeQCv2.4.3_gene_median_tpm.gct.gz", sep="\t", header=2, index_col='Description')
```

```python
df.shape
```

```python
import matplotlib.pyplot as plt
import seaborn
```

```python
thyroid = numpy.log10(df["Thyroid"])
aorta = numpy.log10(df["Artery_Aorta"])
seaborn.histplot([thyroid, aorta], binwidth=1)
```

```python
seaborn.violinplot([thyroid, aorta], color='green', inner='quartile')
```

```python
seaborn.boxplot([thyroid, aorta])
```
